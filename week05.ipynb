```python
import numpy as np

class RNN: # 교재 코드를 참고하였습니다
    def __init__(self, Wx,Wh,b):
        self.params = [Wx,Wh,b]
        self.grads = [np.zeros_like(Wx),np.zeros_like(Wh),np.zeros_like(b)]
        self.cache = None

    def forward(self, x, h_prev):
        Wx,Wh,b = self.params
        t = np.dot(Wh,h_prev) + np.dot(Wx,x) + b
        h_next = np.tanh(t)

        self.cache = (x,h_prev,h_next)
        return h_next


Wx = np.array([[0.1, 0.2],
               [0.3, 0.4]])   
Wh = np.array([[0.5, 0.0],
               [0.0, 0.5]])   
b_h = np.array([[0.1, 0.1]]).T
  
 
# 출력층
W_hy = np.array([0.5, 0.5])  
Why = W_hy.reshape(-1, 1)     
b_y = np.array([0.1])
 
rnn = RNN(Wx, Wh, b_h)
 
xs = [
    np.array([[1.0, 2.0]]).T,  # x_1 
    np.array([[0.0, 1.0]]).T,  # x_2
    np.array([[1.0, 1.0]]).T,  # x_3
]
 
h = np.zeros((2, 1))  # h_0
 
for t, x in enumerate(xs, start=1):
    h = rnn.forward(x, h)
    y = h.T @ Why + b_y
    print(f"t_{t} h={str(h.round(4).T)} y={y.round(4).T}\n")
```
